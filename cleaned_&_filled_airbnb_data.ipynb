```python
def stat(x, df):
    # If x is row index (integer), use iloc
    subset = df.loc[:,x]
    
    mean_val = np.mean(subset)
    var_val  = np.var(subset)
    min_val  = np.min(subset)
    max_val  = np.max(subset)
    
    stats = pd.DataFrame({
        'Mean': [mean_val],
        'Var': [var_val],
        'Min': [min_val],
        'Max': [max_val]
    })
    
    return stats
```
